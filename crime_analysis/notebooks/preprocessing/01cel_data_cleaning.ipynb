# Filtro das cidades

```python
import pandas as pd
```

```python
df = pd.read_csv ('CelularesSubtraidos_2024.csv', sep = ';')
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
df.head(3)
```

```python
print("Total:", len(df))
print("\nRUBRICA:")
print(df['RUBRICA'].value_counts())
```

```python
df['NOME_MUNICIPIO'].value_counts().head(20)
```

```python
df['CIDADE'].value_counts().head(20)
```

```python
df.loc[
    df['NOME_MUNICIPIO'] != df['CIDADE'],
    ['NOME_MUNICIPIO', 'CIDADE']
].value_counts()
```

```python
filtro_sp = df['NOME_MUNICIPIO'].eq('S.PAULO')

df_sp = df.loc[filtro_sp].copy()
```

```python
df_sp.shape
```

```python
df_sp[['LATITUDE', 'LONGITUDE']].isna().sum()
```

```python
df_sp_geo = df_sp.loc[
    df_sp['LATITUDE'].notna() &
    df_sp['LONGITUDE'].notna()
].copy()

df_sp_geo.shape
```

```python
(df_sp_geo['LATITUDE'] == 0).sum()

(df_sp_geo['LONGITUDE'] == 0).sum()
```

```python
df_sp['LATITUDE'] = pd.to_numeric(
    df_sp['LATITUDE'].astype(str).str.replace(',', '.', regex=False),
    errors='coerce'
)

df_sp['LONGITUDE'] = pd.to_numeric(
    df_sp['LONGITUDE'].astype(str).str.replace(',', '.', regex=False),
    errors='coerce'
)
```

```python
df_sp[['LATITUDE', 'LONGITUDE']].dtypes
```

```python
df_sp_geo = df_sp.loc[
    df_sp['LATITUDE'].between(-24, -23) &
    df_sp['LONGITUDE'].between(-47, -46)
].copy()

df_sp_geo.shape
```

```python
df_sp_geo.duplicated(
    subset=['NOME_DELEGACIA', 'ANO_BO', 'NUM_BO']
).sum()
```

```python
df_sp_geo[
    df_sp_geo.duplicated(
        subset=['NOME_DELEGACIA', 'ANO_BO', 'NUM_BO'],
        keep=False
    )
].head(20)
```

```python
df_sp_geo.to_csv(
    'cel_sp_2024_clean.csv', sep = ';',
    index=False
)
```
